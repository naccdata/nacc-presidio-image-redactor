```python
from presidio_analyzer import AnalyzerEngine, PatternRecognizer
from presidio_analyzer.predefined_recognizers.spacy_recognizer import SpacyRecognizer
```

```python
# Create constants
TEXT = ", 4 MARY SMITH: 1231234 10 Aug 00 5:46:48 pm a — 6C2 18Hz Romi General 80dB  51/+1/3/ 4 Gain= 2qB az2 2 = qj = -"

# Create the analyzer engine object
MyAnalyzer = AnalyzerEngine()

# Analyze the text using built in recognizers to the AnalyzerEngine
person_results = MyAnalyzer.analyze(text=TEXT, entities = ["PERSON"], language='en')
date_results = MyAnalyzer.analyze(text=TEXT, entities = ["DATE_TIME"], language='en')
location_results = MyAnalyzer.analyze(text=TEXT, entities = ["LOCATION"], language='en')
phone_results = MyAnalyzer.analyze(text=TEXT, entities = ["PHONE_NUMBER"], language='en')
social_results = MyAnalyzer.analyze(text=TEXT, entities = ["US_SSN"], language='en')
```

```python
# Display of results
print(person_results)
if len(person_results) > 0:
    print("Found person: " + TEXT[person_results[0].start:person_results[0].end])
print(TEXT[person_results[0].start:person_results[0].end])

print(date_results)
if len(date_results) > 0:
    print(TEXT[date_results[0].start:date_results[0].end])

print(location_results)
if len(location_results) > 0:
    print(TEXT[location_results[0].start:location_results[0].end])

print(phone_results)
if len(phone_results) > 0:
    print(TEXT[phone_results[0].start:phone_results[0].end])

print(social_results)
if len(social_results) > 0:
    print(TEXT[social_results[0].start:social_results[0].end])
```

```python
from presidio_analyzer import AnalyzerEngine
from presidio_analyzer.nlp_engine import NlpEngineProvider, TransformersNlpEngine
from fw_presidio_image_redactor.utils.pretrained_transformer import get_pretrained_transformer_model

get_pretrained_transformer_model("obi/deid_roberta_i2b2")

# Define constants
TRANSFORMERS_CONFIG_FILE = (
    "/home/maxthorpe/Git/presidio-image-redactor/fw_presidio_image_redactor/nlp_configs/transformers.yaml"
)
TEXT = ", 4 MARY SMITH: 1231234 10 Aug 00 5:46:48 pm a — 6C2 18Hz Romi General 80dB  51/+1/3/ 4 Gain= 2qB az2 2 = qj = -"
TEXT = TEXT.lower()

# Set pre-trained model as NLP engine
provider = NlpEngineProvider(conf_file=TRANSFORMERS_CONFIG_FILE)
nlp_engine = provider.create_engine()
MyAnalyzerEngine = AnalyzerEngine(nlp_engine=nlp_engine, supported_languages=["en"])

nlp_artifacts = MyAnalyzerEngine.nlp_engine.process_text(TEXT.lower(), "en")
print(nlp_artifacts.entities)
print(nlp_artifacts.scores)
entities = ['PERSON', 'DATE_TIME', 'LOCATION', 'PHONE_NUMBER', 'EMAIL_ADDRESS', 'CREDIT_CARD', 'US_DRIVER_LICENSE', 'US_BANK_NUMBER']

# Analyze the text using built in recognizers to the AnalyzerEngine
person_results = MyAnalyzerEngine.analyze(text=TEXT, entities = entities, language='en', nlp_artifacts=nlp_artifacts)
date_results = MyAnalyzerEngine.analyze(text=TEXT, entities = entities, language='en', nlp_artifacts=nlp_artifacts)
location_results = MyAnalyzerEngine.analyze(text=TEXT, entities = entities, language='en', nlp_artifacts=nlp_artifacts)
phone_results = MyAnalyzerEngine.analyze(text=TEXT, entities = entities, language='en', nlp_artifacts=nlp_artifacts)
social_results = MyAnalyzerEngine.analyze(text=TEXT, entities = entities, language='en', nlp_artifacts=nlp_artifacts)
```
