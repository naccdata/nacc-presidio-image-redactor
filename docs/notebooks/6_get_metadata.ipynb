# Use Presidio to get metadata from dicom file.

```python
import pydicom
from presidio_image_redactor import DicomImageRedactorEngine
```

```python
dcm = pydicom.dcmread("test_images/davidson.dcm")
```

```python
MyEngine = DicomImageRedactorEngine()
original_metadata, is_name, is_patient = DicomImageRedactorEngine._get_text_metadata(dcm)
phi_list = DicomImageRedactorEngine._make_phi_list(original_metadata, is_name, is_patient)
```

```python
print(f"Original metadata: {original_metadata}")
print(f"Is name: {is_name}")
print(f"Is patient: {is_patient}")
print(f"PHI list: {phi_list}")
```

```python
def remove_dicom_metadata(dcm):

    for element in dcm:
        # Save all metadata except the DICOM image itself
        if element.name != "Pixel Data":

            # Track whether this particular element is a name
            if "name" in element.name.lower():
                element.value = "REDACTED"

            # Track whether this particular element is directly tied to the patient
            if "patient" in element.name.lower():
                element.value = "REDACTED"
    
    dcm.save_as("test_images/redacted_metadata.dcm")
                
                
                
            
```

```python
remove_dicom_metadata(dcm)
```
