```python
import pprint
from presidio_analyzer import (
    AnalyzerEngine,
    PatternRecognizer,
    EntityRecognizer,
    Pattern,
    RecognizerResult,
)
from presidio_analyzer.recognizer_registry import RecognizerRegistry
from presidio_analyzer.nlp_engine import NlpEngine, SpacyNlpEngine, NlpArtifacts
from presidio_analyzer.context_aware_enhancers import LemmaContextAwareEnhancer
```

```python
# Define names pattern & constants
names_regex = r"\b([A-Z][a-z]*|[a-z]+|[A-Z]+)(\s([A-Z][a-z]*|[a-z]+|[A-Z]+))*"
TEXT = ", 4 MARY SMITH: 1231234 10 Aug 00 5:46:48 pm a — 6C2 18Hz Romi General 80dB  51/+1/3/ 4 Gain= 2qB az2 2 = qj = -"
```

```python
# Define class and create recognizer
MyAnalyzerEngine = AnalyzerEngine()
names_pattern = Pattern(name="names_pattern", regex=names_regex, score=0.5)
names_recognizer = PatternRecognizer(
    supported_entity="PERSON", patterns=[names_pattern]
)
```

```python
# Test the recognizer
names_result = names_recognizer.analyze(text=TEXT, entities=["PERSON"], nlp_artifacts=None)
for possible_name in names_result:
    print(possible_name)
    print(TEXT[possible_name.start:possible_name.end])
```

```python
# Add the recognizer to the analyzer registry
# MyAnalyzerEngine.registry.add_recognizer(names_recognizer)
results = MyAnalyzerEngine.analyze(
    text=TEXT, entities=["PERSON"], language="en", return_decision_process=True
)
```

```python
# Show the results using the internal decision explanation
pp = pprint.PrettyPrinter()
for result in results:
    print(TEXT[result.start:result.end])
    pp.pprint(result)
    pp.pprint(result.analysis_explanation.__dict__)
    print("\n")
```
